```python
from langchain.tools import tool
from langchain.chat_models import init_chat_model
model = init_chat_model("groq:llama-3.3-70b-versatile")
@tool
def get_weather(location: str) -> str:
    """Get the weather for a given location."""
    return f"The weather at {location} is sunny."
model_with_tools=model.bind_tools([get_weather])
```

```python
response=model_with_tools.invoke("what is the weather of Kathmandu")
response
```

```python
response.tool_calls
```

#tool exeection loop


```python
# model generate tools calls 
message = [{"role": "user", "content": "What is the weather in Kathmandu?"}]
ai_message=model_with_tools.invoke(message)
message.append(ai_message)
for tool_calls in ai_message.tool_calls:
    tool_result=get_weather.invoke(tool_calls)
    message.append(tool_result)
final_reponse=model_with_tools.invoke(message)
print(final_reponse.text)
```
