```python
### Streaming and batch
```

```python
model.invoke("write a essay of 500 words on girl friend")

```

```python
for chunk in model.stream("write a essay of 500 words on girlfriend"):
    print(chunk.text, end="", flush=True)
```

```python
model.invoke("why do parrrot have colourful feathers")
```

```python
for chunk in model.stream("why do parrrot have colourful feathers"):
    print(chunk.text, end="", flush=True)
```
